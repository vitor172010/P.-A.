### Exercicios- dicionario

```python
aluno = [
    "nome": "Ana",
    "idade": 17,
    "curso": "Informática"

]
print(aluno["nome"])
```

```python
aluno = {
    "nome": "Maria",
    "idade": 22,
    "curso": "Engenharia"
}

print(f"Nome: {aluno['nome']}")
print(f"Idade: {aluno['idade']}")
print(f"Curso: {aluno['curso']}")
```

```python
produto = {
    "nome": "Notebook",
    "preco": 2500.00,
    "estoque": 10
}

print("Antes:", produto)
produto["preco"] = 2200.00
print("Depois:", produto)
```

```python
aluno = {
    "nome": "Pedro",
    "idade": 19
}
print("Antes:", aluno)
aluno["curso"] = "Direito"
print("Depois:", aluno)
```

```python
filme = {
    "titulo": "Vingadores",
    "ano": 2019,
    "genero": "Ação",
    "nota": 8.5
}

print(f"Título: {filme['titulo']}")
print(f"Ano: {filme['ano']}")
print(f"Gênero: {filme['genero']}")
print(f"Nota: {filme['nota']}")
```

### Exercicios- conjunto

```python
numeros = {1, 2, 3, 4, 5}
print(numeros)
```

```python
numeros = {1, 2, 2, 3, 3, 3, 4, 5}
print(numeros)

```

```python
frutas = {"maçã", "banana", "laranja"}
frutas.add("morango")
print(frutas)
```

```python
cores = {"vermelho", "azul", "verde", "amarelo"}
cores.remove("verde")
print(cores)
```

```python
nomes = ["Yris", "Luna", "Luna", "Maitê", "Yris", "Ravi", "Gael", "Theo", "Theo"]
nomes_sem_repetir = set(nomes)
print("Como conjunto:", nomes_sem_repetir)
lista_limpa = list(nomes_sem_repetir)
print("Lista sem repetidos:", lista_limpa)
```
