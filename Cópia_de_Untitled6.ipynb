### Exercícios MATC

```python
# 1. Verificar a cor
print("1. Verificar a cor:")
cor = "azul"

match cor:
    case "azul":
        print("Cor azul")
    case _:
        print("Outra cor")

print("-" * 30)
```

```python
# 2. Verificar o número
print("2. Verificar o número:")
numero = 2

match numero:
    case 1:
        print("Um")
    case 2:
        print("Dois")
    case 3:
        print("Três")
    case _:
        print("Número não cadastrado")

print("-" * 30)
```

```python
# 3. Menu de bebidas
print("3. Menu de bebidas:")
opcao = 1

match opcao:
    case 1:
        print("Água")
    case 2:
        print("Suco")
    case 3:
        print("Refrigerante")
    case _:
        print("Opção inválida")

print("-" * 30)
```

```python
# 4. Verificar o final de semana
print("4. Verificar o final de semana:")
dia = "sabado"

match dia:
    case "sabado" | "domingo":
        print("Final de semana")
    case _:
        print("Dia de semana")

print("-" * 30)
```

```python
# 5. Verificar a nota
print("5. Verificar a nota:")
nota = "A"

match nota:
    case "A":
        print("Excelente")
    case "B":
        print("Bom")
    case "C":
        print("Precisa melhorar")
    case _:
        print("Nota inválida")

```

### Exercícios RANGE

```python
# 1. Mostrar os números de 0 até 9
print("1. Números de 0 até 9:")
for numero in range(10):
    print(numero, end=" ")
print("\n" + "-" * 30)
```

```python
# 2. Mostrar os números de 1 até 10
print("2. Números de 1 até 10:")
for numero in range(1, 11):
    print(numero, end=" ")
print("\n" + "-" * 30)
```

```python
# 3. Mostrar os números de 5 até 15
print("3. Números de 5 até 15:")
for numero in range(5, 16):
    print(numero, end=" ")
print("\n" + "-" * 30)
```

```python
# 4. Mostrar os números pares de 0 até 20
print("4. Números pares de 0 até 20:")
for numero in range(0, 21, 2):
    print(numero, end=" ")
print("\n" + "-" * 30)
```

```python
# 5. Mostrar os números ímpares de 1 até 15
print("5. Números ímpares de 1 até 15:")
for numero in range(1, 16, 2):
    print(numero, end=" ")
print("\n" + "-" * 30)
```

```python

# 6. Fazer uma contagem regressiva de 10 até 1
print("6. Contagem regressiva de 10 até 1:")
for numero in range(10, 0, -1):
    print(numero, end=" ")
print("\n" + "-" * 30)
```

```python
# 7. Mostrar a tabuada do número 3
print("7. Tabuada do número 3:")
for i in range(1, 11):
    resultado = 3 * i
    print(f"3 x {i} = {resultado}")
```
