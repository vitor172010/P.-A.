```python
# @title
# Urna Eletrônica
# Criar um programa de votação com 3 candidatos
#Votar candidato A
# Votar candidato B
#Votar candidato C
# Votar Branco
# Votar Nulo
# Encerrar votação
# O sistema deve repetir até encerrar a votação
# Contar votos de cada candidato
# Contar votos brancos e nulos
#Ao encerra, mostrar:
# Total de Votos
# Total de votos por candidato
#Total de votus nulos
#vencedor da eleição
#Regras extras
#se houver empate entre candidatos : mostrar eleição empatada
# se não houver votos válidos,mostrar : "nenhum candidato recebeu votos."

# Inicialização dos contadores

votos_a = 0
votos_b = 0
votos_c = 0
votos_brancos = 0
votos_nulos = 0

while True:
    print("\n--- Urna Eletrônica ---")
    print("1 - Candidato A")
    print("2 - Candidato B")
    print("3 - Candidato C")
    print("4 - Voto Branco")
    print("5 - Voto Nulo")
    print("0 - Encerrar Votação")

    opcao = input("Escolha uma opção: ")

    if opcao == '1':
        votos_a += 1
    elif opcao == '2':
        votos_b += 1
    elif opcao == '3':
        votos_c += 1
    elif opcao == '4':
        votos_brancos += 1
    elif opcao == '5':
        votos_nulos += 1
    elif opcao == '0':
        break
    else:
        print("Opção inválida!")

# Cálculos finais
total_votos = votos_a + votos_b + votos_c + votos_brancos + votos_nulos
votos_validos = [votos_a, votos_b, votos_c]
max_votos = max(votos_validos)

print("\n--- Resultado Final ---")
print(f"Total de Votos: {total_votos}")
print(f"Candidato A: {votos_a}")
print(f"Candidato B: {votos_b}")
print(f"Candidato C: {votos_c}")
print(f"Votos Brancos: {votos_brancos}")
print(f"Votos Nulos: {votos_nulos}")

# Regras Extras
if sum(votos_validos) == 0:
    print("\nNenhum candidato recebeu votos.")
else:
    # Verifica se há empate entre os que tiveram mais votos
    if votos_validos.count(max_votos) > 1:
        print("\nEleição empatada")
    else:
        if votos_a == max_votos:
            print("\nVencedor: Candidato A")
        elif votos_b == max_votos:
            print("\nVencedor: Candidato B")
        else:
            print("\nVencedor: Candidato C")
1

```

```python
# @title

# Jogo Menu inteligente
# Criar um sistema que funcione até o usuário escolher sair.
# Menu
#2 - Positivo / Negativo
#3 Maior
#4 Sair

while True:
    print("\n--- Menu Inteligente ---")
    print("1 - Par / Ímpar")
    print("2 - Positivo / Negativo")
    print("3 - Maior")
    print("4 - Sair")

    opcao = input("\nEscolha uma opção: ")

    if opcao == '1':
        num = int(input("Digite um número: "))
        if num % 2 == 0:
            print(f"O número {num} é PAR.")
        else:
            print(f"O número {num} é ÍMPAR.")

    elif opcao == '2':
        num = float(input("Digite um número: "))
        if num > 0:
            print("O número é POSITIVO.")
        elif num < 0:
            print("O número é NEGATIVO.")
        else:
            print("O número é NEUTRO (Zero).")

    elif opcao == '3':
        n1 = float(input("Digite o primeiro número: "))
        n2 = float(input("Digite o segundo número: "))
        if n1 > n2:
            print(f"O maior é {n1}.")
        elif n2 > n1:
            print(f"O maior é {n2}.")
        else:
            print("Os dois números são iguais.")

    elif opcao == '4':
        print("Saindo do sistema...")
        break

    else:
        print("Opção inválida! Tente novamente.")
```
