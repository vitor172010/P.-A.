```python

# Jogo de adivinhação
# O usuario terá 03 chances para adivinha um número de 0 a 10
# Se o número secreto for maior ou menor que o palpite o programa irá imprimir uma mensagem informando
# Informar se o palpite está quente ou frio.
import random

def jogar():
    numero_secreto = random.randint(0, 10)
    tentativas = 3

    print("### Jogo de Adivinhação ###")
    print("Tente adivinhar o número entre 0 e 10!")

    for i in range(tentativas):
        try:
            palpite = int(input(f"\nTentativa {i+1} de {tentativas}. Qual seu palpite? "))
            print("--------------------------------------------------------------------------------------------")
        except ValueError:
            print("Por favor, digite apenas números inteiros.")
            print("--------------------------------------------------------------------------------------------")
            continue

        if palpite == numero_secreto:
            print("Parabéns! Você acertou!")
            return True # Return True if the game is won

        # Lógica de Maior ou Menor
        if palpite < numero_secreto:
            print("O número secreto é MAIOR que seu palpite.")
        else:
            print("O número secreto é MENOR que seu palpite.")

        # Lógica de Quente ou Frio (distância <= 2 é quente)
        distancia = abs(numero_secreto - palpite)
        if distancia <= 2:
            print("Está QUENTE!")
        else:
            print("Está FRIO!")

    print(f"\nGame Over! O número era {numero_secreto}.")
    return False # Return False if the game is lost

def main():
    while True:
        jogar()
        jogar_novamente = input("Deseja jogar novamente? (s/n): ").lower()
        if jogar_novamente != 's':
            print("Obrigado por jogar!")
            break

main()
```
