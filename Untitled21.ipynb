```python
# Exercício 2 – Contas básicas
# Escreva um programa que:
# • Guarde as notas 7.5, 8.0 e 6.5
# • Calcule a média
# • Mostre a média na tela

nome = str (input("Digite seu nome: "))
nota1 = float (input("digite sua nota do primeiro bimestre:"))
nota2 = float (input("digite sua nota do segundo bimestre:"))
nota3 = float (input("digite sua nota do terceiro bimestre:"))


media = (nota1 + nota2 + nota3) /3

if media >=7:
    print(f"Parabéns {nome}, você foi aprovado com média {media:.2f}!")
elif media >= 5:
    print(f"{nome}, você está de recuperação com média {media:.2f}.")
else:
    print(f"Que pena {nome}, você foi reprovado com média {media:.2f}.")
```

```python
# Exercício 3 – Conversões
# Faça as conversões e mostre o resultado:
# Python
# texto1 = "20"
# texto2 = "3.5"
# numero = 8
# a) Transforme texto1 em inteiro e some com 10 b) Transforme texto2 em real e
# multiplique por 2 c) Transforme número em texto e junte com a palavra "anos"

texto1 =int(input())
texto2 =int(input())
numero =int(input())

# a) Transforme texto1 em inteiro e some com 10
resultado_a = int((texto1) + 10)
print(f"a) O resultado de transformar texto1 em inteiro e somar com 10 é: {resultado_a}")

# b) Transforme texto2 em real e multiplique por 2
resultado_b = float(texto2) * 2
print(f"b) O resultado de transformar texto2 em real e multiplicar por 2 é: {resultado_b}")

# c) Transforme número em texto e junte com a palavra "anos"
resultado_c = str(numero) + " anos"
print(f"c) O resultado de transformar numero em texto e juntar com 'anos' é: {resultado_c}")
```

```python
# Exercício 4 – Expressões aritméticas
# Calcule e mostre o resultado de:
# • O dobro de 15
# • A metade de 90
# • O resto da divisão de 17 por 5
# • 2 elevados à potência 4
# • A média de 10, 8 e 6

num1 =int(input())
num2 =int(input())

dobro = num1 * 2
print(f"O dobro de 15 é: {dobro}")

metade = num2 / 2
print(f"A metade de 90 é: {metade}")

resto_divisao = 17 % 5
print(f"O resto da divisão de 17 por 5 é: {resto_divisao}")

potencia = 2 ** 4
print(f"2 elevados à potência 4 é: {potencia}")

media_numeros = (10 + 8 + 6) / 3
print(f"A média de 10, 8 e 6 é: {media_numeros}")
```
